# Poisson Distribution

Poisson maps continuous time onto discrete counting outcomes, using this formula:-

$$P(X = K) = \frac{e^{-\lambda} \cdot \lambda^{K}}{K!}$$

If we have the rate $r$ and the time interval $t$ we can calculate $\lambda$ as:-

$$\lambda = r \cdot t$$

The expected value of $X$ is $\lambda$ i.e.:-

$$E(X) = \lambda$$

And the expected time $t$ until it reaches count $n$ is:-

$$E(t | X = k) = \frac{k}{r}$$

### Assumptions

Poisson relies on some assumptions. These are:-

- The count is not negative;

- The count is an integer;

- $\lambda$ is constant;

- Events are independent (an event doesn't make another more likely or less likely);

- Mean = $\lambda$;

- Variance = $\lambda$;

### Imports

```python
import numpy as np
```

### Unit Tests

```python
def is_valid_time(x):
    assert(x >= 0)

#Tolerance is used so unit tests dont fail due to floating point rounding errors
def is_valid_probability(x, tolerance = 10 ** -5):
    assert(-tolerance <= x <= 1 + tolerance)
    
def is_valid_probability_distribution(x, tolerance = 10 ** -5):
    assert(1 - tolerance < sum(x) < 1 + tolerance)
    
def is_integer(x):
    assert(x % 1 == 0)
```

### Maths Functions

```python
def factorial(x):
    is_integer(x)
    if x == 0:
        return(1)
    else:
        return(x * factorial(x - 1))
```

### Poisson Class

```python
class Poisson:
    def __init__(self, lamb):
        self.lamb = lamb
    def calculate_prob_x_equals_k(self, k):
        prob = (np.e ** - self.lamb) * (self.lamb ** k) / factorial(k)
        is_valid_probability(prob)
        return(min(prob, 1))
    def calculate_prob_x_less_than_k(self, k):
        prob = sum(self.calculate_prob_x_equals_k(kprime) for kprime in range(k))
        is_valid_probability(prob)
        return(min(prob, 1))
    def calculate_period(self):
        time = 1 / self.lamb
        is_valid_time(time)
        return(time)
    def calculate_expected_time(self, k):
        time = k / self.lamb
        is_valid_time(time)
        return(time)
    def calculate_expected_count(self):
        return(self.lamb)
    def calculate_most_likely_result(self):
        expected = self.calculate_expected_count()
        predicted = int(expected + 0.5)
        is_integer(predicted)
        return(predicted)
    def calculate_probability_distribution(self, tolerance = 10 ** -5): 
        distribution = []
        k = 0
        while sum(distribution) < 1 - tolerance:
            distribution.append(self.calculate_prob_x_equals_k(k))
            k += 1
        is_valid_probability_distribution(distribution, tolerance)
        return(distribution)
    def report_everything(self):
        print(f"Period is: {self.calculate_period():.2f}")
        print(f"Expected count is: {self.calculate_expected_count()}")
        print(f"Most likely result is: {self.calculate_most_likely_result()}")
        distribution = self.calculate_probability_distribution()
        for k in range(len(distribution)):
            print(f"P(X = {k}) = {100 * distribution[k]:.2f}%")
    
model = Poisson(2.7)
model.report_everything()
```
