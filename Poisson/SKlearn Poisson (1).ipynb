# Poisson Distribution

Poisson maps continuous time onto discrete counting outcomes, using this formula:-

$$P(X = K) = \frac{e^{-\lambda} \cdot \lambda^{K}}{K!}$$

If we have the rate $r$ and the time interval $t$ we can calculate $\lambda$ as:-

$$\lambda = r \cdot t$$

The expected value of $X$ is $\lambda$ i.e.:-

$$E(X) = \lambda$$

And the expected time $t$ until it reaches count $n$ is:-

$$E(t | X = k) = \frac{k}{r}$$

### Assumptions

Poisson relies on some assumptions. These are:-

- The count is not negative;

- The count is an integer;

- $\lambda$ is constant;

- Events are independent (an event doesn't make another more likely or less likely);

- Mean = $\lambda$;

- Variance = $\lambda$;

### Imports

```python
import random as r
import numpy as np
from scipy.stats import poisson
from sklearn.linear_model import PoissonRegressor
```

### Data

```python
data_X = np.array([[x] for x in range(10)])

data_Y = np.array([6 * x[0] + r.randint(1, 3) for x in data_X])
```

### Model

Predicted count from Poisson regressor

```python
model = PoissonRegressor()
model.fit(data_X, data_Y)
predicted_count = model.predict([[11]])
print(predicted_count)
```

Probabilities from target value and $\lambda$

```python
x = 0
lamb = 1
probability_of_exactly_x = poisson.pmf(x, mu = lamb)
probability_of_at_most_x = poisson.cdf(x, mu = lamb)
probability_of_at_least_x = 1 + probability_of_exactly_x - probability_of_at_most_x
print(f"Probability x = {x} for lambda = {lamb}: {100 * probability_of_exactly_x:.2f}%")
print(f"Probability x <= {x} for lambda = {lamb}: {100 * probability_of_at_most_x:.2f}%")
print(f"Probability x >= {x} for lambda = {lamb}: {100 * probability_of_at_least_x:.2f}%")
assert(np.fabs(probability_of_at_least_x + probability_of_at_most_x - probability_of_exactly_x) == 1)
```
